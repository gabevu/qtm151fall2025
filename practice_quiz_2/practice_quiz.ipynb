4. Write a function that takes in a single number as an input argument, and returns a single string as the output. The value of the output string is one of two choices, which is determined by whether the input satisfies a logical condition. Then use the .apply() method from Pandas to evaluate this function for every row in a DataFrame. You will be told which column/variable should be input to the function.

```python
import pandas as pd

df = pd.read_csv("HeightWeight-quiz.csv")

# Define function
def weight_category(weight):
    if weight >= 150:
        return "Heavy"
    else:
        return "Light"

# Apply function to Weight_Pounds column
df["Weight_Category"] = df["Weight_Pounds"].apply(weight_category)

# Display results
df.head()
```

5. Given a DataFrame, you will want to randomly assign each row to one of two groups. To do so, you will add a new column to the DataFrame called "random_var" which is filled with random numbers from a given distribution. Then you will create two new DataFrames, based on the whether each row's random variable satisfies a given logical condition. These two DataFrames represent the random assignment into two groups.

```python
import numpy as np
import pandas as pd

df = pd.read_csv("HeightWeight-quiz.csv")

# Add a new random variable column
df["random_var"] = np.random.rand(len(df))  # values between 0 and 1

# Split into two groups based on condition
group_A = df.query("random_var < 0.5")
group_B = df.query("random_var >= 0.5")

print("Group A rows:", group_A.shape[0])
print("Group B rows:", group_B.shape[0])
#💡 You can use np.random.randn() for normal distribution if they specify it.
```

6. Continuing question 5, you will make two scatter plots from the data in each randomly assigned group. So that they appear side-by-side, you will need to use the subplots command from the matplotlib.pyplot library.

```python
import matplotlib.pyplot as plt

# Make subplots side-by-side
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

# Scatter plot for Group A
axes[0].scatter(group_A["Height_Inches"], group_A["Weight_Pounds"])
axes[0].set_title("Group A: Height vs Weight")
axes[0].set_xlabel("Height (inches)")
axes[0].set_ylabel("Weight (pounds)")

# Scatter plot for Group B
axes[1].scatter(group_B["Height_Inches"], group_B["Weight_Pounds"])
axes[1].set_title("Group B: Height vs Weight")
axes[1].set_xlabel("Height (inches)")
axes[1].set_ylabel("Weight (pounds)")

plt.tight_layout()
plt.show()
```
