(2) Aggregate and query

<font size = "3">

Import the data from "sprint_results.csv" and assign it to a DataFrame. Do the following:

- Extract the subset corresponding to 'constructorId == 51' using ```.query()```
- Group by the column "driverId", and then aggregate the maximum over the "laps" column for each driver.
- Extract the subset of drivers with maximum of laps $\ge 20$. Assign this to a DataFrame with a name of your choice.

For full credit, perform all the steps in bullet points above in a **single line** using chaining. (If you use line breaks or wrap in parentheses it doesn't literally need to be one line).

**Hint:** I only see your final submission, not the code you wrote along the way. It might be easier to break it up into multiple lines, then combine when you're done.

````python
# Your answer here

sprint_results = pd.read_csv("data_raw/sprint_results.csv")
sprint_results

sprint_results_answer = (sprint_results
    .query("constructorId == 51") #Step 1
    .groupby("driverId")["laps"].max().reset_index() #step 2
    .query("laps >= 20") #step 3
)

print(sprint_results_answer)
````

(3) Recode a numeric column

<font size = "3">

- Import data from "races.csv" and assign it to a DataFrame
- Recode the "year" column into a new column "year_brackets"
with the following categories

$\qquad$ ``` ["1950-1965","1966-1980","1981-onwards"] ```

The easiest method is ``pd.cut()`` but you are not required to use this.

````python
# Your answer here

races = pd.read_csv("data_raw/races.csv")

races["year_brackets"] = pd.cut(x = races["year"],
                bins=[1950, 1965, 1966, float("inf")],
                labels = ["1950-1965", "1966-1980", "1981-onwards"])
print(races)
````

(4) Aggregate and sort

<font size = "3">

Import the data from "sprint_results.csv" and assign it to a DataFrame (or use the same DataFrame you created in problem (2)). Do the following:

- Group by "driverId", and then aggregate the average (mean) over the "laps" column.
- Sort the teams by average laps in **descending** order using ``.sort_values()``. Assign this to a DataFrame with a name of your choice.

For full credit, perform both steps in bullet points above in a **single line** using chaining. (If you use line breaks or wrap in parentheses it doesn't literally need to be one line).

**Hint:** Same hint as problem 2. You can chain commands together later.

````python
# Your answer here

sprint_results.groupby("driverId").agg(avg_laps = ('laps', 'mean')).sort_values(by = 'avg_laps', ascending = False)
````
